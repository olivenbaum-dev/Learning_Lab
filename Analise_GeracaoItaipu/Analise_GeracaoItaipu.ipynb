```python
import pandas as pd
import matplotlib.pyplot as plt

# Lendo arquivo .csv direto do link
url = "https://ons-aws-prod-opendata.s3.amazonaws.com/dataset/geracao_itaipu/GERACAO_ITAIPU.csv"
df = pd.read_csv(url, sep=';')

# Dica: verifique o formato da data e converta para datetime
df['din_instante'] = pd.to_datetime(df['din_instante'])
df.set_index('din_instante', inplace=True)

# Plot básico para ver a cara dos dados
df['val_itaipu_total'].plot(figsize=(12,6), title="Geração Itaipu ao longo do tempo")
plt.ylabel('Geração Total Itaipu (MW)') # Adicionado rótulo do eixo vertical com unidade
plt.xlabel('Tempo em dias (de 2000 até os dias atuais)') # Adicionado rótulo do eixo horizontal
plt.show()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(14, 10), sharex=True) # Criando 2 subplots, compartilhando o eixo X

# Plot para Itaipu 50Hz
axes[0].plot(df.index, df['val_itaipu_50hz'], label='Itaipu 50Hz (MW)', color='blue')
axes[0].set_title('Geração Itaipu: 50Hz ao Longo do Tempo')
axes[0].set_ylabel('Geração (MW)')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.7)

# Plot para Itaipu 60Hz
axes[1].plot(df.index, df['val_itaipu_60hz'], label='Itaipu 60Hz (MW)', color='red')
axes[1].set_title('Geração Itaipu: 60Hz ao Longo do Tempo')
axes[1].set_xlabel('Data/Hora')
axes[1].set_ylabel('Geração (MW)')
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
